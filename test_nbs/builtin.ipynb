# try except else finally

```python
x = input('Enter a value: ')
```

```python
try:
    y = 1/int(x)

except (ZeroDivisionError, TypeError) as err:
    print(f"Input could not be correctly processed: {err}")

else:
    print(f"{x} transformed into {y}")

finally:
    print("Input processed")
```

# match

```python
import math
```

```python
class Shape:

    def __init__(self):

        print(self.__repr__())

    def description(self):
        print(f"This belongs to Shape parent class")
```

```python
class Circle(Shape):

    def __init__(self, radius):
        # super().__init__()
        self.radius = radius

    def area(self):
        return math.pi * self.radius**2
```

```python
circle = Circle(10)
```

```python
circle.description()
```

```python
circle.area()
```

```python
class Square(Shape):

    def __init__(self, side: int | float):

        self.side = side

    def area(self):
        return self.side**2
```

```python
square = Square(10)
```

```python
square.description()
```

```python
square.area()
```

```python
def match_function(entry: Shape):

    match entry:

        case Circle():
            print(f"radius is {entry.radius}")

        case Square():
            print(f"side is {entry.side}")

        case Shape():
            print(f"{entry.description()}")

        case _:
            print("Not found")
```

```python
match_function(circle)
```

```python
match_function(square)
```

```python
x = Shape()
match_function(x)
```
