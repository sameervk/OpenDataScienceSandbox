```python
"""
Write a function which returns True if it spots the digits 1, 2, 3, and 4 in order in a string.
The digits must appear in order, but other digits can appear between them.
The given string will contain only the digits 0 - 9, and will not contain any duplicated digits.
Therefore, your function should return the following outputs for these cases:
    spot_numbers("1234")     -> True
    spot_numbers("012346")   -> True
    spot_numbers("127354")   -> True
    spot_numbers("01582374") -> True
    spot_numbers("123")     -> False
    spot_numbers("1235678") -> False
    spot_numbers("01243")   -> False
    spot_numbers("4309")    -> False
"""
def spot_numbers(case):
    compare = ['1','2', '3', '4']

    # x = list(case)


    x = ''.join(elem for elem in case if elem in compare)

    
    return '1234' in x      
 
```

```python
spot_numbers("1234")
```

```python
spot_numbers("127354")
```
